# Step 1 : Add required libraries of python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Step 2 : Loading the dataset for in csv file

```python
df=pd.read_csv('/correct_twitter_202102_500MB.csv', encoding='latin1')
```

# Step 3 : Checking the column names

```python
df.head()
```

# Step 4 : Creating a filtered dataset for a searched term such as 'music'

```python
filtered_df = df[df['text'].str.contains('music', case=False).fillna(False)]
```

# Step 5 : To select top 10 dates on which maximum tweets take place.

```python
grouped_df = filtered_df.groupby('created_at').size()
top_10_created_at = grouped_df.nlargest(10)
print(top_10_created_at)
```

# Step 6 : To select number of unique author id

```python
num_unique_authors = filtered_df['author_id'].nunique()
print(num_unique_authors)
```

# Step 7 : To find average of like counts with this term

```python
sum_df = df['like_count'].sum()
sum_term_df = filtered_df['like_count'].sum()
avg = (sum_term_df/sum_df)
print(f"Average percentage: {avg:.2%}")
```

# Step 8 : To select unique place id

```python
num_unique_places = filtered_df['place_id'].unique()
print(num_unique_places)
```

# Step 9 : To select author with maximum tweets.

```python
grouped_df = filtered_df.groupby('author_id').size()
max_tweet = grouped_df.nlargest(1)
print(max_tweet)
```

# Step 10 : Will show the results required

```python
print("\033[1m\033[35mHow many tweets were posted containing the term on each day?(top 10)\033[0m")
print(top_10_created_at )
print("\033[1m\033[35mHow many unique users posted a tweet containing the term?\033[0m")
print(num_unique_authors )
print("\033[1m\033[35mHow many likes did tweets containing the term get, on average?\033[0m")
print(f"Average percentage: {avg:.2%}")
print("\033[1m\033[35mWhere (in terms of place IDs) did the tweets come from?\033[0m")
print(num_unique_places)

print("\033[1m\033[35mWhich user posted the most tweets containing the term?\033[0m")
print(max_tweet)
```
